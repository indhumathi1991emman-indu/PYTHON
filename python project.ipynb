```python
def count_word_in_feedbacks(word):
    count = 0
    word_lower = word.lower()

    for feedback in cleaned_feedbacks:
        if word_lower in feedback.lower():
            count += 1

    return count

print(f"Number of feedbacks containing 'good': {count_word_in_feedbacks('good')}")
print(f"Number of feedbacks containing 'poor': {count_word_in_feedbacks('poor')}")
print(f"Number of feedbacks containing 'excellent': {count_word_in_feedbacks('excellent')}")

```

```python
# 3. Find and display the feedback 
longest_feedback = ""
max_words = 0

for feedback in feedback_data['Feedback']:
    word_count = len(feedback.split())
    if word_count > max_words:
        max_words = word_count
        longest_feedback = feedback

print(f"Longest Feedback (Word: {max_words}):")
print(longest_feedback)
print("\n")
```

```python
# 4. Print the list of unique words
unique_words = set()
for feedback in feedback_data['Feedback']:
    words = feedback.lower().split()
    unique_words.update(words)

print("---Unique Words List---")
print(list(unique_words))
print("\n")
```

```python
# 5. Sort feedback by rating
zipped_feedback = zip(feedback_data['Rating'], feedback_data['Feedback'])

# key=lambda x:[0]
sorted_feedback = sorted(zipped_feedback, key=lambda x: x[0], reverse=True)

print("---Sorted Feedbacks (Highest to Lowest Rating)---")
for rating, feedback in sorted_feedback:
    print(f"Rating: {rating} | Feedback: {feedback}")
```
