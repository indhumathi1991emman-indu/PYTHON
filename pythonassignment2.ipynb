```python
# 1. prompt the user for an input score between 0 and 10
```

```python
score = float(input("Enter your score(0 to 10): "))

# Validate that the score is within the allowed range

if 0 <= score <= 10:
 # 2. Using conditional statements
    if score > 7:
        category = "Above Average"
        feedback = "Excellent job! keep up the fantastic work."
    elif 4 <= score <= 7:
        category = "Average"
        feedback = "Good effort! Keep practicing ,there's room for improvement."
    else:
        category = "Below Average"
        feedback = "Need to improve your performance, consistent practice will lead to better results."
    print(f"{category}: {feedback}")
else:
    print("Invalid input. Please enter a score between 0 and 10.")  
        
```
