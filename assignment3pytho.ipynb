```python
import random
```

```python
# 1. Set up the Game
# Generate random number between 1 and 10
secret_number = random.randint(1, 10)
```

```python
# 2. Prompt the user 
# Set maximum number of attempts allowed
attempts = 3
```

```python
print("Welcome to the Number Guessing Game!")
print("I have selected a number between 1 and 10. Can you guess it?")
print(f"You have {attempts}attempts.\n")
```

```python
while attempts > 0:
    try:
        # Prompt user to guess the game
        guess = int(input(f"Enter your guess (Ateempts left: {attempts}):"))
    except ValueError:
        print("Invalid input! Please enter a valid integer.")
        continue
    if guess < 1 or guess > 10:
        print("Out of bounds! Please guess a number between 1 and 10.")
        continue

    if guess == secret_number:
        print(f"Congratulations! You guessed the correct number: {secret_number}")
        break
    else:
        if guess < secret_number:
            print("Too low!")
        else:
            print("Too high!")

        attempts -= 1
        print("_" * 30)
else:
        print(f"\nGame over! You ran out of attempts.The secret number was {secret_number}. ")
    
```

```python
# 1. Define a function calculate _bmi(weight, height) that returns the BMI
def calculate_bmi(weight, height):
    # Note: BMI = weight (kg) / [height (m)]^2
    bmi = weight / (height **2)
    return bmi

# 2. Prompt the user for their weight (in kg) and height (in meters)
user_weight = float(input("Enter weight in kg:"))
user_height = float(input("Enter your height in meters:"))

# 3.Use the function to calculate and display the BMI
calculated_bmi = calculate_bmi(user_weight, user_height)


print (f"Your BMI is: {calculated_bmi:.2f}")
```
